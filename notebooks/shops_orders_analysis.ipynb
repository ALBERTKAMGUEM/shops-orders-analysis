## Order duration analysis


Restaurant `7965` has a higher price per person than the rest.  
Let's look at the time spent at the table to determine if it matches the characteristics of a fancy restaurant.




Convert the `date_opened` and `date_closed` columns to **datetime** format.


```python
# YOUR CODE HERE
df_order["date_opened"] = pd.to_datetime(df_order["date_opened"])
df_order["date_closed"] = pd.to_datetime(df_order["date_closed"])
```

Run the code below to calculate the time between the opening and closing date of an order in minutes.

```python
# YOUR CODE HERE
df_order["order_duration"] = (df_order["date_closed"] - df_order["date_opened"]).dt.total_seconds() / 60
```

Compute the average duration spent at each table for every restaurant.

```python
# YOUR CODE HERE
df_order.groupby(["id_store"])["order_duration"].mean()
```

Certain restaurants exhibit notably brief average order times, indicating a potential presence of fast food establishments among them.


```python
from nbresult import ChallengeResult

result = ChallengeResult('durations',
                        df_order_shape=df_order.shape,
                        df_order_columns=df_order.columns.tolist(),
                        df_order_types_duration=df_order.dtypes['order_duration'],
)
result.write()
print(result.check())
```

Paris is divided into 20 districts. Each of them has a unique code 75001, 75002, ..., 75019, 75020. \
How many orders were placed in the `2nd district`?

Save the answer to the variable `orders_in_second_district`

```python
# YOUR CODE HERE
df_store.columns
df_merged = df_order.merge(df_store, on="id_store")

orders_in_second_district = (
    df_merged["dim_zipcode"].astype(str) == "75002"
).sum()

print(orders_in_second_district)
```

Let's have a look at restaurant `4542`, which shows a very brief average ordering time.


Retrieve the most ordered items in this restaurant by retaining only the **id_store** and **id_order** columns from `df_order`, as well as the **id_order**, **id_order_line**, and **dim_name** columns from `order_line`.

Store the most ordered item to the variable `most_ordered_at_4542`


```python
# YOUR CODE HERE
# 1. Sélectionner les colonnes demandées
orders = df_order[["id_store", "id_order"]]

order_line = df_line[["id_order", "id_order_line", "dim_name"]]

# 2. Fusionner les deux DataFrames
df_merged = orders.merge(order_line, on="id_order")

# 3. Filtrer le restaurant 4542
df_4542 = df_merged[df_merged["id_store"] == 4542]

# 4. Trouver le produit le plus commandé
most_ordered_at_4542 = df_4542["dim_name"].value_counts().idxmax()

print(most_ordered_at_4542)
```

```python
from nbresult import ChallengeResult

result = ChallengeResult('joins',
                        df_line_shape=df_line.shape,
                        df_line_columns=df_line.columns.tolist(),
                        df_store_shape=df_store.shape,
                        df_store_columns=df_store.columns.tolist(),
                        most_ordered_at_4542=most_ordered_at_4542,
                        orders_in_second_district=orders_in_second_district,
)
result.write()
print(result.check())
```
